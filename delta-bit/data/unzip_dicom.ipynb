```python
#!pip install pydicom nibabel dcmqi
import os, pathlib
import zipfile
import pydicom

NIFTI_FOLDER = pathlib.Path('NIFTI')

def extract_zip(zip_path, extract_to):
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_to)

def convert_dicom_to_nifti(dicom_folder):
    readmeta_data = pydicom.dcmread(dicom_folder / [f for f in os.listdir(dicom_folder) if f.endswith('.dcm')][0])    
    patient_id = readmeta_data.PatientID
    patient_folder = NIFTI_FOLDER / patient_id
    patient_folder.mkdir(exist_ok=True)

    cmd = "dcm2niix -v y -z y -f nativeT1 -o {} {}".format(patient_folder, dicom_folder)
    os.system(cmd)
    return patient_id
```
